```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn.objects as so
```

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python

tips = sns.load_dataset("tips")

sns.scatterplot(data=tips,
                x="total_bill",
                y="tip")

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
sns.scatterplot(data=tips,
                x="total_bill",
                y="tip",
                hue="sex",style="smoker")

plt.show()
```

```python
#line plot 

tips = sns.load_dataset("tips")

sns.lineplot(data=tips,
             x="size",
             y="tip",
             color="blue", marker= "s", markerfacecolor="red",hue="smoker", style="smoker")

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
#barplot

tips = sns.load_dataset("tips")

sns.barplot(data=tips,
            x="day",
            y="tip",
            color="skyblue",palette="Set3",)

plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.show()
```

```python
#count plot 
tips = sns.load_dataset("tips")

sns.countplot(data=tips,
              x="day",
              hue="sex",
              color="skyblue", palette="Set1",
)

plt.title("Count of Days")
plt.xlabel("Day")A
plt.ylabel("Count")
plt.show()
```

```python
tips = sns.load_dataset("tips")

sns.countplot(data=tips,
              x="day",
              hue="smoker",
              color="pink", palette="Set1",
)

plt.title("Count of Days")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

```python
# histogram 

tips = sns.load_dataset("tips")

sns.histplot(data=tips,
             x="total_bill",
             color="skyblue", kde=True)
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.show()
```

```python
tips = sns.load_dataset("tips")

sns.kdeplot(data=tips,
             x="total_bill",
             color="red",
              shade= True)

plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.show()
```

```python
#box plot
# Sample dataset
tips = sns.load_dataset("tips")

# Box Plot
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
#violin plot
tips = sns.load_dataset("tips")

# Box Plot
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
tips = sns.load_dataset("tips")

# Pair Plot
sns.pairplot(tips)

plt.show()
```

```python
#boxen plot
tips = sns.load_dataset("tips")

sns.boxenplot(
    data=tips,
    x="day",
    y="total_bill",
)

plt.title("Boxen Plot")
plt.show()
```

```python
#lmplot
tips = sns.load_dataset("tips")

# LM Plot
sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    col="sex",
    hue="sex",
    palette=["red","orange"]
)

plt.show()
```

```python
#rug plot
tips = sns.load_dataset("tips")

# Rug Plot
sns.rugplot(
    data=tips,
    x="total_bill"
)

plt.title("Rug Plot of Total Bill")
plt.xlabel("Total Bill")
plt.show()
```

```python

tips = sns.load_dataset("tips")

# Rug Plot
sns.ecdfplot(
    data=tips,
    x="total_bill"
)

plt.title("Rug Plot of Total Bill")
plt.xlabel("Total Bill")
plt.show()
```

```python
#The seaborn.objects interface

# sns → old Seaborn functions (scatterplot, lineplot, barplot)
# so → new Objects Interface (Plot, Dot, Line, Bar)
```

```python
#sample Specifying a plot

penguins = sns.load_dataset("penguins")
penguins.head(2)
```

```python
(
    so.Plot(penguins,
            x="bill_length_mm",
            y="bill_depth_mm")
    .add(so.Dot())
)
```

```python
#Setting properties
(
    so.Plot(penguins,
            x="bill_length_mm",
            y="bill_depth_mm")
    .add(so.Dot(color="r", pointsize=5))
)
```

```python
#Mapping properties
#color="species"-----> base to differt differt colur penquin brand
#pointsize="body_mass_g"----> penguin wight based size change 
(
    so.Plot(penguins,
            x="bill_length_mm",
            y="bill_depth_mm", color="species", pointsize="body_mass_g")
    .add(so.Dot())
)
```

```python
#Statistical transformation

(
    so.Plot(penguins, x="species", y="body_mass_g")
    .add(so.Bar(), so.Agg())
)
```

```python
(
    so.Plot(penguins, x="species", y="body_mass_g")
    .add(so.Dot(pointsize=20), so.Agg())
)
```

```python
(
    so.Plot(penguins, x="species", y="body_mass_g",color="sex")
    .add(so.Dot(pointsize=20), so.Agg())
)
```

```python
#Resolving overplotting overlap time ising code 
(
    so.Plot(penguins, x="species", y="body_mass_g", color="sex")
    .add(so.Bar(), so.Agg())
)
```

```python
(
    so.Plot(penguins, x="species", y="body_mass_g",color="sex")
    .add(so.Bar(), so.Agg(),so.Dodge())
)
```

```python
(
    so.Plot(penguins, x="species", y="body_mass_g", color="sex")
    .add(so.Dot(), so.Dodge())
)
```

```python
# #  #Figure-level vs. axes-level functions

# # Axes-level functions create a plot on a single Matplotlib Axes and return an Axes object. They are useful when working with subplots and need more customization.

# # Figure-level functions create the complete figure automatically, can generate multiple subplots, and return a FacetGrid object.
# # Easy way to remember
# Axes-level = One plot 🎯
# Figure-level = Whole figure (one or many plots) 🖼️

# Examples:

# Axes-level: scatterplot(), lineplot(), barplot(), boxplot(), histplot()
# Figure-level: relplot(), catplot(), displot(), pairplot(), jointplot(), lmplot(
```
